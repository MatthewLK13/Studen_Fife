```python
import pandas as pd

# 1. Đọc lại file .csv gốc
file_path = "example-dataset.csv"  # Đổi đường dẫn file tại đây nếu cần
df = pd.read_csv(file_path)
print("Đã tải dữ liệu thành công. Kích thước:", df.shape)
```

```python
# 2. Kiểm tra số lượng dòng trùng lặp hoàn toàn
num_duplicates = df.duplicated().sum()
print(f"Số lượng dòng trùng lặp hoàn toàn: {num_duplicates}")
```

```python
# 3. Kiểm tra số lượng giá trị bị thiếu (missing values) cho từng cột
print("--- Số lượng giá trị bị thiếu từng cột ---")
print(df.isnull().sum())
```

```python
# 4. Kiểm tra các cột dạng số bị nhận diện nhầm thành kiểu 'object' (do chứa khoảng trắng hoặc ký tự lạ)
print("--- Kiểm tra các cột nghi ngờ là số nhưng bị nhận diện nhầm thành kiểu 'object' ---")

found_mistyped = False
for col in df.select_dtypes(include=['object']).columns:
    # Ép thử sang kiểu số
    numeric_series = pd.to_numeric(df[col], errors='coerce')
    valid_numeric_count = numeric_series.notnull().sum()
    
    # Nếu một cột có đa số giá trị (>50%) là số nhưng lại dính một số ô không phải số (khoảng trắng/ký tự lạ)
    if valid_numeric_count > 0.5 * len(df):
        invalid_count = numeric_series.isnull().sum() - df[col].isnull().sum()
        if invalid_count > 0:
            found_mistyped = True
            print(f"⚠️ Cột '{col}': Phát hiện {invalid_count} giá trị bị nhầm lẫn (chứa khoảng trắng hoặc ký tự lạ làm sai kiểu dữ liệu).")
            invalid_values = df[col][numeric_series.isnull() & df[col].notnull()].unique()
            print(f"   -> Giá trị mẫu bị lỗi: {invalid_values[:5]}")

if not found_mistyped:
    print("✅ Không phát hiện cột số nào bị nhầm thành kiểu 'object'. Tất cả các cột 'object' đều là văn bản chuẩn.")
```
