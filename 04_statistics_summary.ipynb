```python
import pandas as pd

# 1. Đọc file -clean.csv vừa làm sạch
file_path = "example-dataset-clean.csv"  # Bạn có thể điều chỉnh tên file nếu cần
df = pd.read_csv(file_path)
print("Đã tải thành công dữ liệu sạch. Kích thước:", df.shape)
df.head()
```

```python
# 2. Thống kê mô tả cơ bản (count, mean, min, max, std,...) cho các cột dạng số
print("--- THỐNG KÊ MÔ TẢ CHO CÁC CỘT DẠNG SỐ ---")
numeric_summary = df.describe()
numeric_summary
```

```python
# 3. Kiểm tra tần suất xuất hiện (value_counts()) của các cột phân loại (categorical/object)
print("--- TẦN SUẤT XUẤT HIỆN CỦA CÁC CỘT PHÂN LOẠI ---")

categorical_cols = df.select_dtypes(include=['object', 'category']).columns

if len(categorical_cols) > 0:
    for col in categorical_cols:
        print(f"\n==========================================")
        print(f"📊 Tần suất xuất hiện cột: [{col}]")
        print("==========================================")
        # Tính số lượng và tỷ lệ phần trăm
        counts = df[col].value_counts()
        percentages = df[col].value_counts(normalize=True) * 100
        
        freq_table = pd.DataFrame({
            'Số lượng (Count)': counts,
            'Tỷ lệ (%)': percentages.round(2)
        })
        print(freq_table)
else:
    print("Không có cột phân loại nào trong tập dữ liệu.")
```
