```python
import pandas as pd

# 1. Đọc file .csv gốc
file_path = "example-dataset.csv"  # Bạn có thể thay bằng đường dẫn file của bạn
df = pd.read_csv(file_path)
print("Đã tải thành công dữ liệu gốc. Kích thước:", df.shape)
```

```python
# 2. Xử lý các lỗi dữ liệu đã phát hiện ở bước trước

# Xóa các dòng trùng lặp hoàn toàn nếu có
initial_rows = len(df)
df = df.drop_duplicates()
print(f"Đã xóa {initial_rows - len(df)} dòng trùng lặp.")

# Tự động phát hiện và ép kiểu các cột số bị chứa khoảng trắng/ký tự lạ
for col in df.select_dtypes(include=['object']).columns:
    # Ép thử sang kiểu số
    converted = pd.to_numeric(df[col], errors='coerce')
    # Nếu số lượng chuyển đổi thành công chiếm phần lớn và có sinh ra NaN mới từ ký tự lạ/khoảng trắng
    if converted.notnull().sum() > 0 and (converted.isnull().sum() - df[col].isnull().sum() > 0):
        print(f"-> Chuyển đổi cột '{col}' sang kiểu số...")
        df[col] = converted
        # Điền giá trị bị thiếu mới tạo bằng giá trị trung vị (median) hoặc 0
        median_val = df[col].median()
        df[col] = df[col].fillna(median_val)
        print(f"   Đã điền các ô bị lỗi/trống trong cột '{col}' bằng giá trị trung vị (median): {median_val}")

# Điền giá trị bị thiếu còn lại cho các cột số khác (nếu có)
for col in df.select_dtypes(include=['number']).columns:
    if df[col].isnull().sum() > 0:
        median_val = df[col].median()
        df[col] = df[col].fillna(median_val)
        print(f"-> Đã điền giá trị thiếu ở cột số '{col}' bằng giá trị trung vị: {median_val}")

# Điền giá trị bị thiếu còn lại cho các cột dạng chữ/chuỗi (nếu có)
for col in df.select_dtypes(include=['object']).columns:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna("Unknown")
        print(f"-> Đã điền giá trị thiếu ở cột chữ '{col}' bằng: 'Unknown'")
```

```python
# 3. Kiểm tra lại lần cuối để đảm bảo kiểu dữ liệu và bảng dữ liệu đã sạch
print("--- KIỂM TRA LẠI KIỂU DỮ LIỆU -----------------")
print(df.dtypes)

print("
--- KIỂM TRA LẠI TỔNG SỐ GIÁ TRỊ THIẾU SE LÀ 0 ---")
print(df.isnull().sum())

print("
Kích thước dữ liệu sau khi làm sạch:", df.shape)
```

```python
# 4. Lưu DataFrame đã làm sạch thành file mới có tên kèm hậu tố -clean.csv (tuyệt đối không ghi đè file gốc)
output_file = file_path.replace(".csv", "-clean.csv")
df.to_csv(output_file, index=False)

print(f"✅ Đã lưu thành công dữ liệu làm sạch vào file mới: {output_file}")
```
